## 1. Introduction to Functions

A function assigns exactly one output to every allowed input. We write $f(x)=x^2$: $x$ is the input and $f(x)$ is the output. For example, $f(3)=9$. A relation is a function only when one input does not produce two different outputs. The vertical line test expresses the same idea graphically: a vertical line may hit a function's graph at most once.

```python
f = x**2
display(Math(rf'f(x) = {sp.latex(f)}'))
display(Math(rf'f(3) = {sp.latex(f.subs(x, 3))}'))
x_values = np.array([-2, -1, 0, 1, 2])
print('Inputs: ', x_values)
print('Outputs:', x_values**2)
```

```python
def numerical_function(expr, variable=x):
    """Turn a SymPy expression into a NumPy-aware function."""
    return sp.lambdify(variable, expr, modules=['numpy'])

def graph_expression(expr, start=-5, stop=5, points=600, title=None, point=None):
    """Plot a SymPy expression and optionally highlight (point, f(point))."""
    grid = np.linspace(start, stop, points)
    values = np.asarray(numerical_function(expr)(grid), dtype=float)
    values[~np.isfinite(values)] = np.nan
    figure = go.Figure(go.Scatter(x=grid, y=values, mode='lines', name='f(x)'))
    if point is not None:
        point_y = float(numerical_function(expr)(point))
        figure.add_trace(go.Scatter(x=[point], y=[point_y], mode='markers+text', text=[f'({point:.2f}, {point_y:.2f})'], textposition='top center', marker={'size': 11, 'color': 'crimson'}, name='selected point'))
    figure.update_layout(title=title or rf'$f(x)={sp.latex(expr)}$', xaxis_title='x', yaxis_title='f(x)', template='plotly_white', height=450)
    figure.update_xaxes(zeroline=True, zerolinecolor='gray')
    figure.update_yaxes(zeroline=True, zerolinecolor='gray')
    return figure

display(graph_expression(f, point=2, title='The graph of $f(x)=x^2$'))
```

## 2. Domain and Range

The **domain** is the set of allowed inputs; the **range** is the set of outputs. For $f(x)=x^2$, the domain is all real numbers and the range is $[0,\infty)$. Restrictions often come from division by zero or even roots of negative numbers. For $g(x)=1/(x-2)$, $x=2$ is excluded. For $h(x)=\sqrt{x+1}$, we require $x\geq-1$.

```python
g = 1 / (x - 2)
h = sp.sqrt(x + 1)
display(Math(rf'g(x) = {sp.latex(g)}'))
display(Math(rf'h(x) = {sp.latex(h)}'))
print('g excludes x=2; h requires x >= -1.')
display(graph_expression(g, start=-5, stop=5, title='Rational function: x=2 is excluded'))
display(graph_expression(h, start=-1, stop=8, title='Square-root function'))
```

Use parentheses for excluded endpoints and brackets for included endpoints. Thus $x\geq-1$ is $[-1,\infty)$, while the domain of $1/(x-2)$ is $(-\infty,2)\cup(2,\infty)$.

## 3. Types of Functions

- **Linear:** $f(x)=mx+b$; constant rate of change.
- **Polynomial:** sums of powers, such as $x^3-2x+1$.
- **Rational:** a quotient of polynomials, such as $1/(x-2)$.
- **Exponential:** $a^x$ for $a>0$.
- **Trigonometric:** periodic functions such as $\sin(x)$ and $\cos(x)$.

```python
function_examples = {'Linear': 2*x + 1, 'Polynomial': x**3 - 2*x + 1, 'Rational': 1/(x - 2), 'Exponential': 2**x, 'Trigonometric': sp.sin(x)}
for name, expression in function_examples.items():
    display(Math(rf'\text{{{name}: }} {sp.latex(expression)}'))

figure = go.Figure()
grid = np.linspace(-4, 4, 600)
for name, expression in function_examples.items():
    values = np.asarray(numerical_function(expression)(grid), dtype=float)
    values[~np.isfinite(values)] = np.nan
    figure.add_trace(go.Scatter(x=grid, y=values, mode='lines', name=name))
figure.update_layout(title='Five function families', xaxis_title='x', yaxis_title='output', template='plotly_white', height=500)
display(figure)
```

## 4. Transformations and Composition

For a base function $f(x)$, $f(x)+k$ shifts up by $k$, $f(x-h)$ shifts right by $h$, $-f(x)$ reflects across the $x$-axis, and $f(-x)$ reflects across the $y$-axis. Composition means putting one function inside another: $(f\circ g)(x)=f(g(x))$.

```python
base = x**2
transformed = 2*(x - 1)**2 - 3
f_base, g_base = x**2, x + 1
composition = sp.expand(f_base.subs(x, g_base))
display(Math(rf'\text{{Base: }} f(x)={sp.latex(base)}'))
display(Math(rf'\text{{Transformation: }} q(x)={sp.latex(transformed)}'))
display(Math(rf'(f\circ g)(x)={sp.latex(composition)}'))
figure = go.Figure()
for name, expression in [('f(x)=x²', base), ('q(x)=2(x−1)²−3', transformed)]:
    figure.add_trace(go.Scatter(x=np.linspace(-4, 4, 500), y=numerical_function(expression)(np.linspace(-4, 4, 500)), mode='lines', name=name))
figure.update_layout(title='A function and its transformation', xaxis_title='x', yaxis_title='output', template='plotly_white')
display(figure)
```

## 5. Reading Graphs

An **x-intercept** occurs where $f(x)=0$; a **y-intercept** is $f(0)$. A local maximum is a nearby high point and a local minimum is a nearby low point. A function is increasing when its outputs rise as $x$ rises. The average rate of change from $a$ to $b$ is $\frac{f(b)-f(a)}{b-a}$, the slope of a secant line.

```python
reading_example = x**3 - 3*x
a, b = -2, 1
average_rate = sp.simplify((reading_example.subs(x, b) - reading_example.subs(x, a)) / (b - a))
roots = sp.solve(reading_example, x)
display(Math(rf'f(x)={sp.latex(reading_example)}'))
display(Math(rf'\text{{x-intercepts: }} {sp.latex(roots)}'))
display(Math(rf'\text{{y-intercept: }} {sp.latex(reading_example.subs(x, 0))}'))
display(Math(rf'\text{{Average rate from }} {a} \text{{ to }} {b}: {sp.latex(average_rate)}'))
display(graph_expression(reading_example, start=-2.2, stop=2.2, point=1, title='Reading features from a graph'))
```

## 6. Mini-project: Interactive Visual Function Explorer

Use the controls to explore function families. The graph, LaTeX formula, selected point, and numerical output update together. Try to predict what will happen before moving a slider. Record one example from each family with its domain, range, intercepts, and shape.

```python
def explorer_expression(family, a, b, c):
    if family == 'Linear': return a*x + c
    if family == 'Quadratic': return a*(x - b)**2 + c
    if family == 'Rational': return a/(x - b) + c
    if family == 'Exponential': return a**(x - b) + c
    return a*sp.sin(x - b) + c

def draw_explorer(family='Quadratic', a=1.0, b=0.0, c=0.0, x0=1.0, xmin=-5.0, xmax=5.0):
    expression = explorer_expression(family, a, b, c)
    function = numerical_function(expression)
    y0 = float(function(x0))
    display(Math(rf'f(x) = {sp.latex(expression)}'))
    display(Markdown(f'**Selected point:** $({x0:.3f}, {y0:.3f})$'))
    grid = np.linspace(xmin, xmax, 700)
    values = np.asarray(function(grid), dtype=float)
    values[~np.isfinite(values)] = np.nan
    figure = go.Figure()
    figure.add_trace(go.Scatter(x=grid, y=values, mode='lines', name='f(x)'))
    figure.add_trace(go.Scatter(x=[x0], y=[y0], mode='markers+text', text=[f'({x0:.2f}, {y0:.2f})'], textposition='top center', marker={'size': 12, 'color': 'crimson'}, name='selected point'))
    figure.update_layout(title=f'{family} function explorer', xaxis_title='x', yaxis_title='f(x)', template='plotly_white', height=500)
    figure.update_xaxes(zeroline=True, zerolinecolor='gray')
    figure.update_yaxes(zeroline=True, zerolinecolor='gray')
    display(figure)

family_widget = widgets.Dropdown(options=['Linear', 'Quadratic', 'Rational', 'Exponential', 'Sine'], value='Quadratic', description='Family:')
a_widget = widgets.FloatSlider(value=1.0, min=-3.0, max=3.0, step=0.1, description='a:', continuous_update=False)
b_widget = widgets.FloatSlider(value=0.0, min=-4.0, max=4.0, step=0.1, description='b:', continuous_update=False)
c_widget = widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.1, description='c:', continuous_update=False)
x0_widget = widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.1, description='x₀:', continuous_update=False)
xmin_widget = widgets.FloatSlider(value=-5.0, min=-10.0, max=0.0, step=0.5, description='x min:', continuous_update=False)
xmax_widget = widgets.FloatSlider(value=5.0, min=0.0, max=10.0, step=0.5, description='x max:', continuous_update=False)
controls = widgets.VBox([family_widget, a_widget, b_widget, c_widget, x0_widget, xmin_widget, xmax_widget])
output = widgets.interactive_output(draw_explorer, {'family': family_widget, 'a': a_widget, 'b': b_widget, 'c': c_widget, 'x0': x0_widget, 'xmin': xmin_widget, 'xmax': xmax_widget})
display(widgets.HBox([controls, output]))
```
